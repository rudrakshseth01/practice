```python
x=df[['Weight']]
type(x)
# independent variable should be kept in dataframe or in 2d array so that it is stored in two cols nd importance is given to the data
y=df['Height']#dependent var can be stored in series or one d arra
```

```python
# trin test split 
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.25,random_state=10)
```

```python
x_test.shape
```

```python
x_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scalar = StandardScaler()
x_train=scalar.fit_transform(x_train)
# x_train
```

```python
x_test=scalar.transform(x_test)
# x_test
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regression=LinearRegression(n_jobs=-1)
```

```python
regression.fit(x_test,y_test)
```

```python
print('coefficient : ',regression.coef_)
print('intercept : ',regression.intercept_)
```

```python
# plottting best fit line for taining data
plt.scatter(x_train,y_train)
plt.plot(x_train,regression.predict(x_train))
```

## prediction of test data
predicted height ouput= itercept +coeff.weights

```python
y_pred=regression.predict(x_test)
```

```python
from sklearn.metrics import *
```

```python
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
rmse=np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

```python
score=r2_score(y_test,y_pred)
score
```

## Adjusted R2 = 1 - [(1-R2)*(n-1)/(n-k-1)]

n: The number of observations
